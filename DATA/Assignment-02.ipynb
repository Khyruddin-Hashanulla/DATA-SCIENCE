```python
import json

# Load the data
def load_data(filename):
  with open(filename, "r") as f:
      data = json.load(f)
  return data 

data = load_data("StoreData.json")

# Clean and Structure the data
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users=set()
    
    for user in data:
        # Clean Ratings - Data consistency
        raw_rating = user["rating"].strip().lower()
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating

        # Handle Missing Value
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None

        # Deduplication
        if(user["name"].strip() in unique_users):
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data

data = clean_data(data)
```

```python
# Get meaningful insights from data
def get_insights(data):

    # Average Rating
    total_rating = 0

    for user in data:
        total_rating += float(user["rating"])

    print(f"Average Rating: {total_rating / len(data)}")


    # Percentage of users with poor rating
    poor_rating = 0

    for user in data:
        if float(user["rating"]) < 3:
            poor_rating += 1

    print(f"Percentage of users with poor rating: {poor_rating / len(data) * 100}%")
```

```python
get_insights(data)
```
